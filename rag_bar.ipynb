## 1. El modelo no conoce el bar

### Ejercicios

**1.1** Pregúntale al modelo, sin el manual, a qué hora cierra la cocina del bar La Consulta el domingo.

Resultado esperado: se lo inventa o dice que no lo sabe

**1.2** Busca en el manual el párrafo que lo dice y pásaselo al modelo con la pregunta.

Resultado esperado: a las 15:45

## 2. De texto a números

### Ejercicios

**2.1** Compara "¿Dónde dejo un paraguas olvidado?" con "objetos perdidos" y con "la wifi de los clientes". ¿Cuál se parece más?

Resultado esperado: objetos perdidos, 0.84 (la wifi, 0.75)

**2.2** Compara "El bar abre el lunes" con "El bar cierra el lunes". ¿Se parecen más o menos que las frases del 2.c?

Resultado esperado: 0.96, más que "fatal" y "enfermo"

**2.3** ¿Cuántos números tiene el vector del manual entero?

Resultado esperado: 384

## 3. Trocear el manual

### Ejercicios

**3.1** ¿Qué trozos hablan de propinas?

Resultado esperado: uno, el 24

**3.2** ¿Cuál es el trozo más largo, y cuántas letras tiene?

Resultado esperado: el 22, con 321 letras

**3.3** ¿Qué trozo contiene "En ese caso"? ¿Se entiende solo?

Resultado esperado: el 18, y no dice de qué caso habla

## 4. Buscar los trozos que contestan

### Ejercicios

**4.1** Busca qué dice el manual de los perros, con la pregunta "¿Pueden entrar perros en el bar?".

Resultado esperado: el 15 primero, con 0.86

**4.2** Busca "¿Dónde dejo un paraguas olvidado?". ¿Comparte alguna palabra con el trozo que sale primero?

Resultado esperado: el 35, sin ninguna palabra en común

**4.3** Busca algo que el manual no dice: "¿Cuánto se facturó el sábado 7 de noviembre?". ¿Salen trozos? ¿Cuánto se parece el primero?

Resultado esperado: salen tres igualmente; el primero, 0.80, menos que el 0.86 de los perros

**4.4** Busca "¿Me puedo tomar una cerveza cuando acabo?" por parecido, y después saca los trozos que tienen la palabra "cerveza". ¿Coinciden?

Resultado esperado: por parecido, el 12 (el del alcohol); con la palabra, el 34 (el del grifo)

## 5. El prompt y la respuesta

### Ejercicios

**5.1** Cambia las instrucciones para que conteste como si fuera Rosa, de tú y en una frase, y vuelve a preguntar.

**5.2** Pregunta "¿Qué vino le recomiendo a un cliente para acompañar el pulpo?" con unas instrucciones que solo digan quién es, sin las dos frases de cómo contestar. ¿De dónde saca la respuesta?

Resultado esperado: recomienda vinos que no salen en el manual

## 6. Todo en una función, y un chat

### Ejercicios

**6.1** Hazle al chat tres preguntas que estén en el manual y una que no.

**6.2** Ponle al chat tres preguntas de ejemplo, de las que se pulsan.

**6.3** Haz otra función, preguntar_con_trozos, que añada al final de la respuesta los números de los trozos que ha usado.

Resultado esperado: la respuesta, y debajo algo como (Trozos: 22 5 25)

## 7. Cuando falla

### Ejercicios

**7.1** Busca la pregunta "Un cliente es celíaco, ¿qué le puedo poner?". ¿Alguno de los trozos que salen habla del gluten de los platos?

Resultado esperado: salen los trozos 31, 25 y 29, y el primero es el de las quejas

**7.2** Pregunta "¿El bar abre los lunes?" y "¿El bar cierra los lunes?". ¿Contesta bien a las dos?

Resultado esperado: sí: el lunes está cerrado
