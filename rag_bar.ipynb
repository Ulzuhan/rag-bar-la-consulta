## 2. De texto a números

En la web: pestaña Parecidos. Si al cargar el modelo sale un aviso de HF_TOKEN, no hay que hacer nada.

**2.a** El modelo que convierte textos en números.

**2.b** Un texto, 384 números.

**2.c** Cuánto se parecen dos frases.

### Ejercicios

**2.1** Compara "¿Dónde dejo un paraguas olvidado?" con "objetos perdidos" y con "la wifi de los clientes". ¿Cuál se parece más?

## 4. Buscar los trozos que contestan

En la web: pestaña Buscar.

**4.a** Todos los trozos, en números.

**4.b** La pregunta, también.

**4.c** Los que más se parecen.

**4.d** La búsqueda, en una función.

### Ejercicios

**4.1** Busca qué dice el manual de los perros, con la pregunta "¿Pueden entrar perros en el bar?".

Resultado esperado: el 15 primero, con 0.87

**4.2** Busca "¿Dónde dejo un paraguas olvidado?". ¿El trozo que sale primero dice "paraguas" u "olvidado"?

Resultado esperado: no: sale el 35, el de objetos perdidos, que no dice ninguna de las dos

**4.3** Busca algo que el manual no dice: "¿Cuánto se facturó el sábado 7 de noviembre?". ¿Salen trozos? ¿Cuánto se parece el primero?

Resultado esperado: salen tres de todas formas; el primero, con 0.81, casi como el 0.82 del paraguas, que sí era el bueno

**4.4** Busca "¿Me puedo tomar una cerveza cuando acabo?" por parecido, y después saca los trozos que tienen la palabra "cerveza". ¿Coinciden?

Resultado esperado: por parecido, el 12 (el del alcohol); con la palabra, el 34 (el del grifo)
